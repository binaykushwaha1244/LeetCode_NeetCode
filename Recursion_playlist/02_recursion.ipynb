```python
#functional way of doing recursion
# sum of first N numbers
def sum(n):
    if n == 0:
        return 0
    return n + sum(n-1)

n = 5
print(sum(n))
```

```python
# Parameterised way of doing recursion

def sum_i(i, sum):
    if i < 1:
        print(sum)
        return
    sum_i(i-1, sum+i)

n = 5
print(sum_i(n,0))

sum_i(n,0)
```

```python
# Functional way 
# factorial of n

def fact(n):
    if n == 0:
        return 1
    return n * fact(n-1)

n = 5
print(fact(n))

# Time and Space Complexity: O(n),    stored in Stack spaces
```

```python
# parameterised Way of recursion

def fact(i, prod):
    if i == 0:
        print(prod)
        return
    fact(i-1, prod*i)

n = 5
fact(n, 1)
```

```python
# reverse an array

arr = [1,2,3,4,2]
rev = []

for i in range(len(arr)-1,-1,-1):
    rev.append(arr[i])

print(rev)
```
