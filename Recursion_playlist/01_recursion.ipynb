```python
def name_print(name, n):
    if n == 0:
        return True

    print(name)
    name_print(name, n-1)

name = "Binay 8.8 cgpa"
n = 4

name_print(name, n)
```

```python
def name_copy(i, n):
    if i > n:
        return True
    print("binay 8.8 cgpa")
    name_copy(i+1,n)

n = 4
name_copy(0,n)
```

```python
# print linearly 1 to N  N = 4. out: 1 2 3 4 

def num(i,n):
    if i > n:
        return True

    print(i)
    num(i+1,n)

n = 5
num(1,n)
```

```python
# print from N to 1

def revNum(i, n):
    if i > n:
        return True

    print(n)
    revNum(i, n-1)

n = 5
revNum(1,n)
```

```python
#print linearly from 1 to n using backtrack  NOT f(i+1,n)

def fun(i,n):
    if i<1:
        return
    fun(i-1,n)
    print(i)

n = 4
fun(n,n)
```

```python
# Print from N to 1 using backtrack 

def revnum(i,n):
    if i > n:
        return
    revnum(i+1,n)
    print(i)

n = 4
revnum(1,4)
```
